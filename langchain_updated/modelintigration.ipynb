```python
from langchain.chat_models import init_chat_model
model=init_chat_model('gpt-4.1')
model
```

```python
# To use Llama 3.3 70B (Recommended)
model = init_chat_model("groq:llama-3.3-70b-versatile")
response = model.invoke("why do parrots speak?")
print(response.content)
```

## Streaming

### it is used to generate content on realtime , as and when the content is generated it will start showing up.

```python
for chunk in model.stream("why so parrot speak?"):
    print(chunk.text, end="|",flush=True)
```

## Batch


```python
response=model.batch(
    [ "why do parrot speak",
    " how many states were ther in America at first",
    "How many presidents have served US"],
    config={
        'max_concurrency': 5,
    }
)
```
