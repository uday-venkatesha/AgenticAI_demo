```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name: str
    age: int
    city: str


person = Person(name="Alice", age=30, city="New York")
print(person)
```

```python
#lets try with wrong data type
person = Person(name="Alice", age="thirty", city="New York")
print(person)
```

#### Gives an error , this is why pydantic is usefull , it is extensiovely used for data validation , especially when we do API call , it validates , if we are getting the right data type.

### Lests try with the tradition Dataclass


```python
from dataclasses import dataclass
@dataclass
class PersonDC:
    name: str
    age: int
    city: str
person_dc = PersonDC(name="Alice", age="thirty", city="New York")
print(person_dc)


# it works without any error , this is why pydantic is usefull , it is extensiovely used for 
# data validation , especially when we do API call , it validates , if we are getting the right data type.
# In contrast, traditional dataclasses do not perform any validation, which can lead to runtime errors if the 
# data does not conform to the expected types.
```

```python
## add Optional fields using python optional type
from typing import Optional
class employee(BaseModel):
    name: str
    age: int
    city: str
    department: Optional[str] = None  # Optional field with default value None
    is_active: Optional[bool] = True  # Optional field with default value True
    
```

```python
employee1 = employee(name="Bob", age=25, city="Los Angeles", department="Engineering")
print(employee1)
```

```python

## Nested Models

from pydantic import BaseModel
class address(BaseModel):
    street: str
    city: str
    zip_code: str

class customer(BaseModel):
    name: str
    age: int
    address: address  # Nested Pydantic model

customer1 = customer(
    name="Charlie",
    age=28,
    address={
        "street": "123 Main St",
        "city": "Springfield",
        "zip_code": "12345"
    }
)
print(customer1)
```

## Pydantic field 
## - The field function in pydantic enhances model field beyond basic type , hints by allowing you to specify validation rules , default alues , alias,and more 

```python
from pydantic import Field, BaseModel

class Product(BaseModel):
    name: str = Field(..., title="Product Name", max_length=100)
    price: float = Field(..., gt=0, description="Price must be greater than zero")
    in_stock: bool = Field(default=True, description="Availability of the product")
    
product1 = Product(name="Laptop", price=999.99)
```

```python
print(product1)
```
